## Message的调用

```python
import os

from fastapi import FastAPI
from langchain_core.messages import HumanMessage, AIMessage,SystemMessage
from langchain.chat_models import init_chat_model
from dotenv import load_dotenv
from rich import print as rprint

load_dotenv(override=True)

deepseek = init_chat_model(
    model_provider="deepseek",
    api_key=os.getenv("AI_API_KEY"),
    model="deepseek-flash",
    base_url=os.getenv("AI_BASE_URL"),
    temperature=0.7,
    max_tokens=1024,
)
```

```python

messages = [
    SystemMessage(content="You are a helpful assistant.And you shall response in chinese"),
    HumanMessage("What is the meaning of life?"),
    AIMessage('This is a test.'),
    HumanMessage("What did i just ask?")
]
deepseek.invoke(messages).pretty_print()
```

# 多轮对话机器人

```python
from langchain.chat_models import init_chat_model
from dotenv import load_dotenv
import os

load_dotenv(override=True)

MAX_PAIRS = 3
EXIT_WORD = "exit"



deepseek = init_chat_model(
    model_provider="deepseek",
    api_key=os.getenv("AI_API_KEY"),
    model="deepseek-flash",
    base_url=os.getenv("AI_BASE_URL"),
    temperature=0.7,
    max_tokens=1024,
)

messages = [
    {'role': 'system', 'content': 'You are a helpful assistant.Named dawg'}
]

print('当输入 exit 时退出对话')

turn = 1


def keep_recent_messages(messages,max_pairs):
    system_messages = [m for m in messages if m.get('role') == 'system']
    conversation_message = [m for m in messages if m.get('role') != 'system']

    recent_messages = conversation_message[-(max_pairs*2):]

    return system_messages + recent_messages

while True:
    print('\n','='*10,f'第{turn}轮对话开始','='*10,'\n')
    user_input = input("请输入你的问题：")
    if user_input == EXIT_WORD:
        print("退出对话")
        break

    messages.append({'role': 'user', 'content': user_input})
    response = deepseek.invoke(messages)

    memory_messages = keep_recent_messages(messages, max_pairs=MAX_PAIRS)

    reply_content = ''

    for chunk in deepseek.stream(memory_messages):
        if chunk.content:
            print(chunk.content, end='', flush=True)
            reply_content += chunk.content

    print('\n','='*10,f'第{turn}轮对话结束','='*10,'\n')
    turn += 1

    messages.append({'role':'assistant','content':reply_content})
```
