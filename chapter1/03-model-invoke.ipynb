# 关于模型调用invoke的使用

## 1 invoke的传参
### 1.1文本的输入

```python
import os
from langchain.chat_models import init_chat_model
from dotenv import load_dotenv
from langchain_core.messages import SystemMessage, AIMessage

load_dotenv(override=True)
AI_API_KEY = os.getenv("AI_API_KEY")

AI_BASE_URL = os.getenv("AI_BASE_URL")

deepseek = init_chat_model(
    api_key=AI_API_KEY,
    model='deepseek-flash',
    base_url=AI_BASE_URL,
    temperature=0.7,
    max_tokens=1024
)

conversation = [
    {'role': "system", 'content': "你是一名专业的心理医生导师"},
    {'role': "user", 'content': "为什么人会有嫉妒的心理，怎么解决"},
]
response3 = deepseek.invoke(conversation)
print('AI responses:', response3.content)

conversation.append({'role':'assistant','content':response3.content})
conversation.append({'role':'user','content':'what did i just ask'})
response4 = deepseek.invoke(conversation)
print(response4.content)
```

传递记忆

```python
import os
from langchain.chat_models import init_chat_model
from dotenv import load_dotenv

load_dotenv(override=True)
AI_API_KEY = os.getenv("AI_API_KEY")

AI_BASE_URL = os.getenv("AI_BASE_URL")

deepseek = init_chat_model(
    api_key=AI_API_KEY,
    model='deepseek-flash',
    base_url=AI_BASE_URL,
    temperature=0.7,
    max_tokens=1024
)

conversation = [
    {'role': "system", 'content': "你是一名专业的心理医生导师"},
    {'role': "user", 'content': "为什么人会有嫉妒的心理，怎么解决"},
]
response3 = deepseek.invoke(conversation)
print('AI responses:', response3.content)

conversation.append({'role':'assistant','content':response3.content})
conversation.append({'role':'user','content':'what did i just ask'})
response4 = deepseek.invoke(conversation)
print(response4.content)
```

传递记忆

```python
from langchain_core.messages import HumanMessage
from langchain_core.messages import SystemMessage
from langchain_core.messages import AIMessage


message4 = [
    SystemMessage(content='你是一名专业的心理医生导师'),
    HumanMessage(content='为什么人会有嫉妒的心理，怎么解决')
]
a = deepseek.invoke(message4).content
message4.append(AIMessage(content=a))
print(deepseek.invoke(message4).content)
message4.append(HumanMessage(content='what did i just ask'))
print(deepseek.invoke(message4).content)
```
