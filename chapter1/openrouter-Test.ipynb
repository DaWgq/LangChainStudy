```python
import os
from langchain.chat_models import init_chat_model
from dotenv import load_dotenv

load_dotenv(override=True)
AI_API_KEY = os.getenv("openrouter_key")

AI_BASE_URL = os.getenv("openrouter_url")

openrouter = init_chat_model(
    model_provider="openrouter",
    api_key=AI_API_KEY,
    model='nvidia/nemotron-3.5-lightning:free',
    base_url=AI_BASE_URL,
    temperature=0.7,
    max_tokens=1024,
    max_retries = 3
)
prompt = '人生的意义是什么'
response = openrouter.stream(prompt)
for i in response:
    print(i.text,end='',flush=True)
```
